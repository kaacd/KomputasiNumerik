```python
import numpy as np

# Contoh Sistem Persamaan Linear 3 Variabel:
# 3x + 4y - z = 12
# 6x + 6y - 4z = 8
# -3x + 5y - 12z = 6

# Matriks awal (tipe data float agar pembagian desimal akurat)
matriks_spltv = np.array([
    [3.0,  4.0, -1.0, 12.0],
    [6.0,  6.0, -4.0, 8.0],
    [-3.0, 5.0, -12.0, 6.0]
])

print("Matriks Augmented Awal:")
print(matriks_spltv)
```

### 1. Inisialisasi Pustaka (Library) dan Penentuan Matriks SPL
Pada cell ini, kita mengimpor pustaka `numpy` yang sangat efisien untuk melakukan operasi matriks dan vektor di Python.
- `np.array`: Digunakan untuk mendefinisikan matriks augmented $A|b$ dari Sistem Persamaan Linear (SPL) tiga variabel yang akan diselesaikan.
- Tipe data diatur ke float (`.0`) agar hasil pembagian desimal selama proses komputasi numerik terjaga ketelitiannya.

```python
# Salin matriks agar data awal tetap aman
M = matriks_spltv.copy()
n = len(M)

for i in range(n):
    # Partial Pivoting sederhana: pastikan elemen diagonal utama tidak nol
    if M[i][i] == 0.0:
        for k in range(i + 1, n):
            if M[k][i] != 0.0:
                M[[i, k]] = M[[k, i]]
                break

    # Proses eliminasi baris di bawah pivot
    for j in range(i + 1, n):
        faktor = M[j][i] / M[i][i]
        M[j] = M[j] - faktor * M[i]

print("Matriks Setelah Eliminasi Maju (Segitiga Atas):")
print(np.round(M, 4))
```

### 2. Proses Eliminasi Maju (Forward Elimination)

Tujuan dari tahap ini adalah mengubah matriks augmented awal menjadi bentuk **Matriks Segitiga Atas** (semua elemen di bawah diagonal utama bernilai nol). Ini adalah mesin utama program kita. Fungsi gauss_eliminasi bekerja dalam beberapa langkah:

- **Partial Pivoting Sederhana:** Mengecek apakah elemen diagonal utama bernilai nol (`0.0`). Jika ya, baris akan ditukar dengan baris di bawahnya untuk menghindari pembagian dengan angka nol (`ZeroDivisionError`).
- **Faktor Pengali (Multiplier):** Dihitung dengan rumus $m_{ji} = a_{ji} / a_{ii}$, lalu baris di bawahnya diperbarui dengan mengurangkan baris tersebut dengan perkalian faktor pengali dan baris pivot.
- `np.round(M, 4)`: Digunakan untuk membulatkan hasil matriks hingga 4 angka dibelakang koma agar tampilan matriks lebih rapi dan mudah dibaca.

- Matriks Augmented: Menggabungkan persamaan di sisi kiri ($A$) dan sisi kanan ($b$) menjadi satu tabel besar (M) agar mudah dihitung sekaligus.

- Eliminasi Maju: Proses ini membuat bentuk "segitiga bawah" matriks menjadi angka nol.

- Partial Pivoting: Sebelum baris dikurangi, program akan mencari angka terbesar di kolom tersebut dan menukarnya ke posisi atas (pivot). Ini mencegah error pada komputer (menghindari pembagian dengan angka yang mendekati nol).

- Substitusi Mundur: Setelah matriks berbentuk seperti tangga, program mencari nilai variabel dari yang paling bawah (misalnya $z$), lalu naik ke atas untuk mencari $y$, dan terakhir mencari $x$.

- Membuat Nol (Eliminasi): Ini adalah inti dari metode Gauss. Kita ingin membuat angka di bawah pivot menjadi  0 .

- Substitusi Naik: Setelah nilai variabel paling bawah ketemu, kita "naik" ke baris atasnya.

```python
# Array kosong untuk menampung hasil x, y, z
solusi = np.zeros(n)

# Menghitung dari baris terakhir ke baris pertama
for i in range(n - 1, -1, -1):
    jumlah_kanan = 0.0
    for j in range(i + 1, n):
        jumlah_kanan += M[i][j] * solusi[j]
    solusi[i] = (M[i][n] - jumlah_kanan) / M[i][i]

print("Hasil Penyelesaian SPL:")
print(f"Nilai x = {solusi[0]:.4f}")
print(f"Nilai y = {solusi[1]:.4f}")
print(f"Nilai z = {solusi[2]:.4f}")
```

### 3. Proses Substitusi Balik (Back Substitution) dan Hasil Akhir
Setelah matriks berbentuk segitiga atas didapatkan, tahap terakhir adalah mencari nilai variabel ($x, y, z$) dari baris terbawah ke atas:
- `np.zeros(n)`: Membuat array kosong berukuran $3$ untuk menampung nilai solusi akhir.
- **Looping Mundur (`range(n - 1, -1, -1)`):** Menghitung nilai variabel mulai dari baris terakhir (variabel terakhir) ke baris pertama. Nilai variabel yang sudah diketahui disubstitusikan ke persamaan di atasnya untuk mencari variabel lainnya.
- Menampilkan hasil akhir nilai variabel $x$, $y$, dan $z$ dengan format desimal 4 angka penting.
