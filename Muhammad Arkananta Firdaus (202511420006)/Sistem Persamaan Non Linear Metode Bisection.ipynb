```python
import numpy as np

# Definisi fungsi non-linear: f(x) = x^3 - 3x - 1
def fungsi(x):
    return x**3 - 3*x - 1
```

### 1. Inisialisasi Pustaka dan Definisi Fungsi Persamaan Non-Linear
Pada bagian awal program ini, kita menyiapkan pustaka (*library*) Python dan menetapkan persamaan matematika yang akan diselesaikan.
* **`import numpy as np`**: Berfungsi untuk memuat pustaka NumPy ke dalam program. NumPy sangat penting dalam komputasi numerik karena menyediakan fungsi matematika cepat dan penanganan tipe data angka yang akurat.
* **Definisi Fungsi `fungsi(x)`**: Kita membuat fungsi matematika non-linear, yaitu $f(x) = x^3 - 3x - 1$, menggunakan keyword `def` di Python. Tujuan utama kita dalam metode biseksi ini adalah mencari nilai $x$ tertentu (akar persamaan) di mana hasil akhir fungsi tersebut bernilai nol ($f(x) = 0$).

```python
def metode_biseksi(a, b, toleransi, max_iterasi):
    # Validasi awal: pastikan fungsi(a) dan fungsi(b) memiliki tanda berlawanan
    if fungsi(a) * fungsi(b) >= 0:
        raise ValueError("Interval awal tidak valid! fungsi(a) dan fungsi(b) harus memiliki tanda berbeda.")

    riwayat = []

    for i in range(1, max_iterasi + 1):
        # Hitung titik tengah (c)
        c = (a + b) / 2.0
        nilai_fungsi_c = fungsi(c)

        # Simpan data iterasi ke dalam list riwayat
        riwayat.append([i, a, b, c, nilai_fungsi_c])

        # Cek kriteria pemberhentian (stopping criteria)
        if abs(nilai_fungsi_c) < toleransi or (b - a) / 2.0 < toleransi:
            break

        # Perbarui batas interval
        if fungsi(a) * nilai_fungsi_c < 0:
            b = c
        else:
            a = c

    return c, riwayat
```

### 2. Implementasi Logika Utama `metode_biseksi`
Cell ini memuat algoritma inti metode biseksi (bagi dua) untuk mempersempit rentang tebakan secara berulang:
* **Validasi Awal (`fungsi(a) * fungsi(b) >= 0`)**: Memeriksa apakah interval awal yang dimasukkan benar-benar mengapit akar (tanda nilai fungsi di titik $a$ dan $b$ harus berlawanan). Jika tandanya sama, program akan memunculkan pesan error.
* **Pencarian Titik Tengah (`c = (a + b) / 2.0`)**: Menghitung titik tengah dari interval aktif dan mengevaluasi nilainya menggunakan variabel `nilai_fungsi_c`.
* **Perekaman Riwayat (`riwayat.append`)**: Menyimpan setiap tahapan proses perulangan dari awal hingga akhir ke dalam daftar list.
* **Kriteria Berhenti & Pembaruan Interval**: Perulangan akan berhenti jika tingkat ketelitian (`toleransi`) sudah tercapai, atau menggeser batas interval $a$ dan $b$ menuju titik tengah $c$ berdasarkan tanda hasil kali fungsi.

```python
# Menentukan batas interval awal dan parameter error
batas_a = -2.0
batas_b = 3.0
toleransi_error = 0.0001
batas_iterasi = 50

try:
    # Memanggil fungsi metode biseksi
    akar_persamaan, data_iterasi = metode_biseksi(batas_a, batas_b, toleransi_error, batas_iterasi)

    print("=== TABEL DETAIL PROSES ITERASI METODE BISECTION ===")
    print(f"{'Iterasi':<8} | {'a':<10} | {'b':<10} | {'c (Tengah)':<12} | {'fungsi(c)':<10}")
    print("-" * 60)
    for row in data_iterasi:
        print(f"{row[0]:<8} | {row[1]:<10.4f} | {row[2]:<10.4f} | {row[3]:<12.4f} | {row[4]:<10.4f}")

    print("\n" + "=" * 50)
    print(f"Akar persamaan ditemukan pada x = {akar_persamaan:.5f}")
    print(f"Total iterasi yang dilakukan   : {len(data_iterasi)}")

except ValueError as err:
    print(f"Terjadi Kesalahan: {err}")
```
