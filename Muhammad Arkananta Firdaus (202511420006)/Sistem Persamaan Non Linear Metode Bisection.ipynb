```python
import numpy as np
import pandas as pd
```

```python
def bisection_method(func, a, b, tol=1e-6, max_iter=100):
    """
    Menyelesaikan persamaan non-linear f(x) = 0 menggunakan Metode Bisection.

    Parameter:
    - func     : Fungsi matematika f(x) yang akan dicari akarnya
    - a        : Batas bawah interval awal
    - b        : Batas atas interval awal
    - tol      : Toleransi galat error yang diizinkan (default 1e-6)
    - max_iter : Batas maksimum iterasi

    Returns:
    - c        : Estimasi nilai akar persamaan
    - history  : List berisi kamus data proses iterasi untuk pelacakan tabel
    """

    # 1. Validasi awal berdasarkan Teorema Bolzano:
    # Pastikan f(a) dan f(b) memiliki tanda berlawanan (berbeda tanda)
    if func(a) * func(b) >= 0:
        raise ValueError("Interval awal [a, b] tidak valid: f(a) dan f(b) harus memiliki tanda berlawanan (f(a) * f(b) < 0).")

    history = []

    # 2. Proses perulangan iterasi numerik
    for i in range(1, max_iter + 1):
        # Menghitung titik tengah (c) dari interval [a, b]
        c = (a + b) / 2.0
        f_c = func(c)
        f_a = func(a)

        # Menghitung galat estimasi (setengah dari panjang interval saat ini)
        galat = abs(b - a) / 2.0

        # Menyimpan data riwayat iterasi untuk analisis tabel
        history.append({
            'Iterasi': i,
            'a': a,
            'b': b,
            'c (Titik Tengah)': c,
            'f(c)': f_c,
            'Galat': galat
        })

        # 3. Kondisi Berhenti (Konvergensi)
        # Jika nilai fungsi di c sama dengan 0 atau galat sudah lebih kecil dari toleransi
        if f_c == 0.0 or galat < tol:
            return c, history

        # 4. Menentukan sub-interval baru berdasarkan tanda nilai fungsi
        if f_a * f_c < 0:
            b = c  # Akar berada di sebelah kiri [a, c]
        else:
            a = c  # Akar berada di sebelah kanan [c, b]

    print("Peringatan: Jumlah iterasi maksimum tercapai sebelum konvergensi penuh.")
    return c, history
```

  #### Menyelesaikan persamaan non-linear f(x) = 0 menggunakan Metode Bisection.

    Parameter:
    - func     : Fungsi matematika f(x) yang akan dicari akarnya
    - a        : Batas bawah interval awal
    - b        : Batas atas interval awal
    - tol      : Toleransi galat error yang diizinkan (default 1e-6)
    - max_iter : Batas maksimum iterasi

    Returns:
    - c        : Estimasi nilai akar persamaan
    - history  : List berisi kamus data proses iterasi untuk pelacakan tabel


```python
# ==========================================
# DEFINISI FUNGSI NON-LINEAR f(x)
# ==========================================
# Contoh fungsi: f(x) = x^3 - 3x - 1
# Kita ingin mencari nilai x di mana f(x) = 0
def f(x):
    return x**3 - 3*x - 1

# Menentukan batas interval awal [a, b] dan parameter toleransi
a_input = -2.0
b_input = 3.0
toleransi_input = 1e-5
maks_iterasi_input = 50

print("Memulai proses komputasi Metode Bisection...")
print("=" * 60)

try:
    # Memanggil fungsi bisection_method yang telah dibuat di cell sebelumnya
    akar_persamaan, riwayat_iterasi = bisection_method(
        f, a_input, b_input, tol=toleransi_input, max_iter=maks_iterasi_input
    )

    print("\n=== HASIL AKHIR PENYELESAIAN ===")
    print(f"Akar persamaan ditemukan pada x = {akar_persamaan:.6f}")
    print(f"Total iterasi yang dilakukan   : {len(riwayat_iterasi)}")

except ValueError as err:
    print(f"Terjadi Kesalahan: {err}")
```

```python
# Mengonversi list riwayat iterasi menjadi Pandas DataFrame agar tampil rapi
df_hasil = pd.DataFrame(riwayat_iterasi)

print("=== TABEL DETAIL PROSES ITERASI METODE BISECTION ===")
# Mengatur format tampilan angka desimal agar mudah dibaca
pd.set_option('display.float_format', lambda x: '%.6f' % x)
display(df_hasil)
```
